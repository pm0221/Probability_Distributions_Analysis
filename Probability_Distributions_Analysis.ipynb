```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import bernoulli

p = 0.39
students = 1000

data = bernoulli.rvs(p, size=students)

placed = np.sum(data)
not_placed = students - placed
print("Placed:", placed)
print("Not Placed:", not_placed)
print("Placement Percentage:", (placed/students)*100, "%")

plt.hist(data, bins=2, edgecolor='black', color='hotpink')
plt.xticks([0, 1], ["Not Placed", "Placed"])
plt.xlabel("Placement_Status")
plt.ylabel("No's of Students")
plt.title("Bernoulli Dist – Placement Status (Histogram)")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import binom
import matplotlib.pyplot as plt
from scipy.stats import binom

n = 500
p = 0.55

data = binom.rvs(n, p, size=100)

plt.hist(data, bins=15, edgecolor='white', color='lightgreen')
plt.title("Binomial Distribution OF STUDENT PLACED")
plt.xlabel("Number of Students Placed")
plt.ylabel("Frequency")
plt.show()
```

```python
import matplotlib.pyplot as plt
from scipy.stats import poisson

mu = 3

data = poisson.rvs(mu, size=100)

plt.hist(data, color='yellow')
plt.title("Poisson Distribution – Job Offers")
plt.xlabel("Number of Job Offers")
plt.ylabel("Frequency")
plt.show()
```

```python
import matplotlib.pyplot as plt
from scipy.stats import bernoulli, binom, poisson, norm, expon, uniform
mean = 70
std = 10

marks = norm.rvs(mean, std, size=100)

plt.hist(marks, bins=10, edgecolor='black', density=True, color="skyblue")
plt.title("Normal Distribution – Student Marks")
plt.xlabel("Marks")
plt.ylabel("Density")
plt.show()
```

```python
mean = 70
std = 10

marks = norm.rvs(mean, std, size=100)

plt.hist(marks, bins=10,edgecolor='black', density=True, color="blue")
plt.title("Normal Distribution – Student Marks")
plt.xlabel("Marks")
plt.ylabel("Density")
plt.show()
```

```python
scale = 5

waiting_time = expon.rvs(scale=scale, size=100)

plt.hist(waiting_time, bins=10,edgecolor='black', density=True, color="orange")
plt.title("Exponential Distribution – Waiting Time")
plt.xlabel("Time")
plt.ylabel("Density")
plt.show()
```

```python
low = 0
high = 100

uniform_data = uniform.rvs(loc=low, scale=high-low, size=100)

plt.hist(uniform_data, bins=10,edgecolor='white', density=True, color="purple")
plt.title("Uniform Distribution – Random Values")
plt.xlabel("Value")
plt.ylabel("Density")
plt.show()
```
